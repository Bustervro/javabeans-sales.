# JavaBeans Coffee Sales Analysis

This project uses Pandas to analyze coffee sales from 2021 and 2022. I will calculate total sales, yearly growth, and find the best and least-selling drinks.

## Part 1: Create the DataFrame

```python
import pandas as pd

# Create coffee sales data
data = {
    'Americano': [5628, 6239],
    'Latte': [12426, 25384],
    'Espresso': [852, 756],
    'Cappuccino': [8522, 7561]
}

# Create the DataFrame
coffee_sales = pd.DataFrame(
    data,
    index=['2021 Sales', '2022 Sales']
)

print('JavaBeans Coffee Sales:')
display(coffee_sales)
```

## Part 2: Calculate Total Sales

```python
# Calculate total sales for each beverage
total_sales = coffee_sales.sum(axis=0)

print('Total Sales by Beverage:')
display(total_sales)
```

## Part 3: Year-over-Year Growth

```python
# Calculate growth from 2021 to 2022
growth = (
    (coffee_sales.loc['2022 Sales'] - coffee_sales.loc['2021 Sales'])
    / coffee_sales.loc['2021 Sales']
) * 100

print('Year-over-Year Growth (%):')
display(growth.round(2))
```

## Part 4: Best and Least-Selling Drinks

```python
# Find the best and least-selling drinks
for year in coffee_sales.index:
    best = coffee_sales.loc[year].idxmax()
    least = coffee_sales.loc[year].idxmin()

    print(f'{year}:')
    print(f'Best-selling drink: {best}')
    print(f'Least-selling drink: {least}')
    print()
```

## Part 5: Business Insights and Strategy

Latte was the best-selling drink in both 2021 and 2022. Its sales increased by about 104.28%. This shows strong customer demand. JavaBeans should spend more money promoting Latte because it is growing quickly. Americano sales also increased by about 10.86%, making it another good drink to promote.

Espresso was the least-selling drink in both years. Espresso sales dropped by about 11.27%, while Cappuccino sales dropped by about 11.28%. JavaBeans should find out why these drinks are losing sales. The company could try new flavors, discounts, or better advertising before deciding whether to remove them from the menu.
